### Teste do Qui-Quadrado   (não paramétrico)
O teste qui-quadrado é uma técnica estatística utilizada para verificar se existe uma associação significativa entre duas
variáveis categóricas (qualitativas). Ele avalia se a distribuição das frequências observadas em uma tabela de contingência
difere significativamente das frequências esperadas, assumindo que não há associação entre as variáveis.

Esse teste é amplamente usado para analisar tabelas de contingência, onde são comparadas as frequências observadas
de diferentes categorias de duas variáveis. O resultado nos permite inferir se há uma relação estatisticamente significativa
entre as categorias das variáveis analisadas

```python
from cinza.Estats import *
df = pd.read_excel("data/qui-quadrado.xlsx")
df.head()
```

```python
r = teste_qui_quadrado(df, "Faixa de Renda", "Tipo de Conteúdo", rp_abs_min=1.9)
```

```python
r["resultado"]
```

Nosso p-valor do qui-quadrado é inferior à 5% (alfa), logo rejeitamos a hipotese nula
- H0: as categorias são independentes
- H1: as categorias NÃO são independentes

```python
r["tabela"]
```

É possível observar, a partir dos resultados apresentados, que:

* Indivíduos com faixa salarial anual entre 75.000 e 99.999 tendem a apresentar maior preferência por filmes;
* Pessoas com renda anual inferior a 25.000 tendem a preferir séries;
* Essa tendência também é observada entre indivíduos com faixa salarial anual entre 25.000 e 49.999, que demonstram maior preferência por séries.

Conclui-se, portanto, que indivíduos pertencentes às menores faixas salariais apresentam uma tendência maior de preferência por séries, enquanto faixas salariais mais elevadas demonstram maior preferência por filmes.

### Teste do sinal (não paramétrico)
O teste utilizado para comparar amostras ordinais pareadas é o teste do sinal. Esse teste é uma versão não paramétrica,
pois lida com variáveis ordinais, onde não é possível assumir normalidade.

Dados ordinais são categorizados em uma ordem (por exemplo, "Ruim", "Bom", "Muito bom", "Excelente"), mas as
diferenças entre as categorias são difíceis de quantificar matematicamente. Isso ocorre porque, em dados ordinais, a
magnitude da diferença entre as categorias não é necessariamente uniforme ou conhecida.

Nesses casos, utilizamos o teste do sinal para verificar se existe uma diferença significativa na população, observando se
as direções das diferenças entre os pares de observações sugerem uma mudança consistente.

```python
from cinza.Estats import *
df = pd.read_excel("data/teste_do_sinal.xlsx")
df.head()
```

Testando se a versão NOVA é "melhor" que a versão antiga

```python
teste_do_sinal(df, ["Nota_Versao_Antiga", "Nota_Versao_Nova"], tipo=">", alfa=0.05)
```

Testando se a versão ANTIGA é "melhor" que a versão nova

```python
teste_do_sinal(df, ["Nota_Versao_Antiga", "Nota_Versao_Nova"], tipo="<", alfa=0.05)
```

Testando se existe diferença entre as duas versões (!=)

```python
teste_do_sinal(df, ["Nota_Versao_Antiga", "Nota_Versao_Nova"], tipo="!=", alfa=0.05)
```

### Teste do McNemar (não paramétrico)
Para comparar as proporções de duas amostras pareadas, utilizamos o teste de McNemar. Esse teste é ideal para avaliar
se há diferença significativa entre as proporções de dois grupos em amostras pareadas, sendo muito comum em análises
de "antes e depois" ou quando os sujeitos são emparelhados.

Hipóteses do Teste de McNemar:
* Ho: As proporções de sucessos são iguais entre os grupos.
* Ha: As proporções de sucessos são diferentes entre os grupos.

        Amostras pareadas (proporções)
        Variaveis Binarias
        tabela de contigencia (2x2)


```python
from cinza.Estats import *
df = pd.read_excel("data/dados_mcnemar_politica.xlsx")
df.head()
```

```python
# teste de McNemar (sem correção de continuidade)
r = teste_de_mcNemar(df, ["Antes_Politica", "Depois_Politica"], alfa=0.05, correcao=False)
```
