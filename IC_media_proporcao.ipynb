### Intervalo de confiança para a média

##### Explicação


O intervalo de confiança para a média é uma técnica de inferência estatística utilizada para estimar um intervalo de valores plausíveis para a média populacional $\mu$ a partir de uma amostra.

Ao invés de utilizar apenas a média amostral $\bar{x}$ como estimativa pontual, o intervalo de confiança fornece uma faixa de valores que provavelmente contém a verdadeira média da população.

#### Fórmula

Quando a variância populacional é desconhecida, utiliza-se a distribuição t de Student:

$$
IC = \bar{x} \pm t_{\alpha/2,n-1}
\left(
\frac{s}{\sqrt{n}}
\right)
$$

Onde:

- $\bar{x}$: média amostral;
- $s$: desvio padrão amostral;
- $n$: tamanho da amostra;
- $t_{\alpha/2,n-1}$: valor crítico da distribuição t de Student;
- $\frac{s}{\sqrt{n}}$: erro padrão da média.

#### Componentes do Intervalo

##### Média Amostral

Representa a melhor estimativa pontual para a média populacional.

$$
\bar{x}
$$

##### Erro Padrão

Mede a variabilidade da média amostral.

$$
EP =
\frac{s}{\sqrt{n}}
$$

Quanto maior a amostra, menor tende a ser o erro padrão.

##### Margem de Erro

Representa a distância entre a estimativa pontual e os limites do intervalo.

$$
ME =
t_{\alpha/2,n-1}
\times
EP
$$

##### Limites do Intervalo

Limite inferior:

$$
LI =
\bar{x}
-
ME
$$

Limite superior:

$$
LS =
\bar{x}
+
ME
$$

#### Interpretação

Suponha que foi obtido:

$$
IC_{95\%}
=
[45,2 ; 51,8]
$$

A interpretação correta é:

> Com 95% de confiança, a média populacional é compatível com valores entre 45,2 e 51,8.

Também pode ser escrito como:

> A média populacional foi estimada em 48,5 unidades, com margem de erro de 3,3 unidades e nível de confiança de 95%.

#### Nível de Confiança

O nível de confiança não representa a probabilidade da média estar dentro do intervalo calculado.

A interpretação correta é:

> Se inúmeras amostras fossem coletadas e um intervalo de confiança fosse construído para cada uma delas, aproximadamente 95% desses intervalos conteriam a verdadeira média populacional.

#### Aplicações

O intervalo de confiança para a média é amplamente utilizado para:

- Estimar salários médios;
- Estimar produtividade média;
- Estimar consumo médio;
- Avaliar desempenho médio de processos;
- Analisar indicadores de negócios e pesquisa científica.

### Intervalo de confiança para proporção

##### Explicação


O intervalo de confiança para a proporção é uma técnica de inferência estatística utilizada para estimar um intervalo de valores plausíveis para uma proporção populacional $p$ a partir de uma amostra.


É frequentemente utilizado quando a variável de interesse possui apenas dois resultados possíveis, como:

- Sim ou Não;
- Sucesso ou Fracasso;
- Comprou ou Não Comprou;
- Aprovado ou Reprovado.

## Estimativa da Proporção

A proporção amostral é calculada por:

$$
\hat{p}
=
\frac{x}{n}
$$

Onde:

- $x$: quantidade de sucessos observados;
- $n$: tamanho da amostra.

##### Fórmula

O intervalo de confiança para a proporção é dado por:

$$
IC
=
\hat{p}
\pm
z_{\alpha/2}
\sqrt{
\frac{\hat{p}(1-\hat{p})}{n}
}
$$

Onde:

- $\hat{p}$: proporção amostral;
- $n$: tamanho da amostra;
- $z_{\alpha/2}$: valor crítico da distribuição normal padrão;
- $\sqrt{\frac{\hat{p}(1-\hat{p})}{n}}$: erro padrão da proporção.

#### Componentes do Intervalo

##### Proporção Amostral

Representa a melhor estimativa pontual da proporção populacional.

$$
\hat{p}
=
\frac{x}{n}
$$

##### Erro Padrão

Mede a variabilidade da proporção amostral.

$$
EP
=
\sqrt{
\frac{\hat{p}(1-\hat{p})}{n}
}
$$

##### Margem de Erro

Representa a distância entre a estimativa pontual e os limites do intervalo.

$$
ME
=
z_{\alpha/2}
\times
EP
$$

##### Limites do Intervalo

Limite inferior:

$$
LI
=
\hat{p}
-
ME
$$

Limite superior:

$$
LS
=
\hat{p}
+
ME
$$

##### Interpretação

Suponha que uma pesquisa encontrou:

$$
\hat{p}
=
0,58
$$

com intervalo de confiança de 95%:

$$
IC_{95\%}
=
[0,48 ; 0,68]
$$

A interpretação correta é:

> Com 95% de confiança, a proporção populacional é compatível com valores entre 48% e 68%.

Também pode ser escrita como:

> A proporção populacional foi estimada em 58%, com margem de erro de aproximadamente 10 pontos percentuais e nível de confiança de 95%.

##### Nível de Confiança

O nível de confiança refere-se à confiabilidade do procedimento utilizado para construir o intervalo.

A interpretação correta é:

> Se inúmeras amostras fossem coletadas e um intervalo de confiança fosse construído para cada uma delas, aproximadamente 95% desses intervalos conteriam a verdadeira proporção populacional.

##### Aplicações

O intervalo de confiança para a proporção é frequentemente utilizado para:

- Pesquisas eleitorais;
- Pesquisas de satisfação;
- Taxa de aprovação;
- Taxa de conversão de vendas;
- Taxa de defeitos em processos produtivos;
- Estudos epidemiológicos e científicos.
