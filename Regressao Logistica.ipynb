# Regressão Logística

### Regressão Logística Binária

##### Explicação

A Regressão Logística Binária é uma técnica estatística utilizada para modelar a relação entre uma variável resposta binária e um conjunto de variáveis explicativas.

A variável resposta assume apenas dois resultados possíveis, como:

- Aprovação (1) ou Reprovação (0)
- Compra (1) ou Não Compra (0)
- Inadimplente (1) ou Adimplente (0)
- Doente (1) ou Saudável (0)

O principal objetivo do modelo é estimar a **probabilidade de ocorrência do evento de interesse**.

---

## Por que não utilizar Regressão Linear?

Se utilizássemos uma regressão linear tradicional para estimar probabilidades, poderiam surgir valores inválidos:

- Probabilidades menores que 0
- Probabilidades maiores que 1

Como probabilidades devem estar sempre no intervalo:

$$
0 \le P(Y=1) \le 1
$$

a regressão logística utiliza uma transformação matemática capaz de restringir as previsões ao intervalo entre 0 e 1.

---

## Função Logística

A regressão logística modela a probabilidade através da função logística:

$$
P(Y=1|X)=\frac{e^{\beta_0+\beta_1X_1+\cdots+\beta_kX_k}}
{1+e^{\beta_0+\beta_1X_1+\cdots+\beta_kX_k}}
$$

onde:

- $P(Y=1|X)$ = probabilidade do evento ocorrer;
- $\beta_0$ = intercepto;
- $\beta_i$ = coeficientes do modelo;
- $X_i$ = variáveis explicativas.

Essa função produz uma curva em formato de "S" (sigmoide), garantindo que as probabilidades previstas permaneçam entre 0 e 1.

---

## O Conceito de Odds (Chance)

Antes de interpretar os coeficientes da regressão logística, é necessário compreender o conceito de **Odds**.

Odds representam a razão entre:

- a probabilidade do evento ocorrer;
- a probabilidade do evento não ocorrer.

Matematicamente:

$$
Odds = \frac{P}{1-P}
$$

### Exemplo

Suponha que a probabilidade de um cliente realizar uma compra seja de 80%:

$$
P=0,80
$$

Logo:

$$
Odds=\frac{0,80}{1-0,80}
=\frac{0,80}{0,20}
=4
$$

Interpretação:

> As chances de compra são 4 vezes maiores que as chances de não compra.

---

## O Conceito de Log-Odds

A regressão logística não modela diretamente a probabilidade.

Ela modela o logaritmo natural das Odds:

$$
\log\left(\frac{P}{1-P}\right)
=
\beta_0+\beta_1X_1+\cdots+\beta_kX_k
$$

Essa transformação recebe o nome de:

- Logit
- Log-Odds
- Logaritmo das Chances

A principal vantagem é transformar um problema não linear em uma relação linear entre os preditores e a resposta.

---

## Equação do Modelo Logístico

O modelo estimado é:

$$
\log\left(\frac{P}{1-P}\right)
=
\beta_0+\beta_1X_1+\cdots+\beta_kX_k
$$

Interpretação:

Cada coeficiente $\beta_i$ representa a variação esperada no **Log-Odds** quando a variável correspondente aumenta uma unidade, mantendo as demais constantes.

Embora matematicamente correta, essa interpretação costuma ser pouco intuitiva.

Por isso, normalmente convertemos os coeficientes para Odds Ratios.

---

## Odds Ratio (OR)

Aplicando a exponencial ao coeficiente:

$$
OR=e^{\beta_i}
$$

obtemos o **Odds Ratio**, que possui interpretação muito mais prática.

O Odds Ratio indica quanto as chances do evento mudam quando a variável aumenta uma unidade.

---

## Interpretação do Odds Ratio

### Caso 1: OR > 1

Suponha:

$$
\beta=0,50
$$

Então:

$$
OR=e^{0,50}=1,65
$$

Interpretação:

> Cada aumento de uma unidade na variável aumenta as chances do evento em aproximadamente 65%.

---

### Caso 2: OR < 1

Suponha:

$$
\beta=-0,70
$$

Então:

$$
OR=e^{-0,70}=0,50
$$

Interpretação:

> Cada aumento de uma unidade na variável reduz as chances do evento em aproximadamente 50%.

---

### Caso 3: OR = 1

Suponha:

$$
\beta=0
$$

Então:

$$
OR=e^0=1
$$

Interpretação:

> A variável não possui efeito sobre as chances do evento.

---

## Variáveis Dummies

Quando o modelo possui variáveis categóricas, são criadas variáveis indicadoras (dummies).

Exemplo:

| Modalidade | Home Office | Presencial |
|------------|------------|------------|
| Home Office | 0 | 0 |
| Presencial | 1 | 1 |

Considerando Home Office como categoria de referência:

$$
OR=e^{\beta}
$$

Interpretação:

- OR > 1 → categoria possui maiores chances que a referência;
- OR < 1 → categoria possui menores chances que a referência.

---

## Teste de Hipótese dos Coeficientes

Para cada variável é realizado o teste:

### Hipótese Nula

$$
H_0:\beta_i=0
$$

### Hipótese Alternativa

$$
H_1:\beta_i\neq0
$$

Se:

$$
p\text{-value}<0,05
$$

concluímos que a variável possui influência estatisticamente significativa sobre a probabilidade do evento.

---

## Interpretação Completa de um Coeficiente

Suponha que o modelo produziu:

$$
\beta_{Experiencia}=0,15
$$

Calculando o Odds Ratio:

$$
OR=e^{0,15}=1,162
$$

Interpretação:

> Mantidas constantes as demais variáveis do modelo, cada ano adicional de experiência aumenta as chances do evento ocorrer em aproximadamente 16,2%.

---

## Probabilidade Prevista

Após estimar os coeficientes, podemos calcular a probabilidade prevista para qualquer observação:

$$
P(Y=1)=
\frac{e^{z}}
{1+e^{z}}
$$

onde:

$$
z=\beta_0+\beta_1X_1+\cdots+\beta_kX_k
$$

Essa probabilidade será sempre um valor entre 0 e 1.

---

## Regra de Classificação

Após obter a probabilidade estimada, define-se um ponto de corte.

O mais comum é:

$$
0,5
$$

Assim:

- Se $P(Y=1)\ge 0,5$ → Classe 1
- Se $P(Y=1)<0,5$ → Classe 0

O ponto de corte pode ser ajustado conforme os objetivos do problema.

---

## Resumo

A Regressão Logística Binária:

- Modela probabilidades de eventos binários;
- Utiliza a função logística para garantir probabilidades entre 0 e 1;
- Modela o Log-Odds do evento;
- Permite interpretar os coeficientes através do Odds Ratio;
- Aceita variáveis quantitativas e qualitativas (dummies);
- É amplamente utilizada em áreas como marketing, medicina, crédito, recursos humanos e ciência de dados.

A interpretação prática dos resultados normalmente é feita através do **Odds Ratio**, obtido pela exponenciação dos coeficientes:

$$
OR=e^{\beta}
$$

pois ele informa diretamente o quanto as chances do evento aumentam ou diminuem quando uma variável sofre alteração.

##### Resultados da Regressão Logística

| Variável               | Coeficiente (Log-Odds) | Odds Ratio | p-value |
| ---------------------- | ---------------------: | ---------: | ------: |

##### Interpretação dos Coeficientes

O coeficiente associado ao **Tempo de Assinatura** foi positivo e estatisticamente significativo (p = 0,00223). Seu Odds Ratio igual a **1,17222** indica que, para cada mês adicional de assinatura, as chances de churn aumentam aproximadamente **17,2%**, mantendo as demais variáveis constantes.

Por outro lado, o coeficiente associado às **Horas Assistidas por Mês** foi negativo e altamente significativo (p < 0,001). O Odds Ratio de **0,93791** sugere que cada hora adicional assistida reduz as chances de churn em aproximadamente **6,2%**.

Esses resultados indicam que clientes mais engajados com a plataforma tendem a apresentar menor propensão ao cancelamento.

##### Estimativa para um Cliente Específico

Deseja-se estimar a probabilidade de churn para um cliente que:

* Está assinando o serviço há **24 meses**;
* Assistiu **32 horas** de conteúdo no último mês.

A equação logística estimada foi:

$$
\text{logit}(p)=0,46498+0,15890(\text{Tempo})-0,06410(\text{Horas})
$$

Substituindo os valores observados:

$$
\text{logit}(p)=0,46498+0,15890(24)-0,06410(32)
$$

$$
\text{logit}(p)=2,22738
$$

Convertendo o logit para probabilidade:

$$
p=\frac{e^{2,22738}}{1+e^{2,22738}}
$$

$$
p \approx 0,903
$$

---

###### Conclusão

Com base no modelo estimado, um cliente que possui **24 meses de assinatura** e assistiu **32 horas de conteúdo no último mês** apresenta uma probabilidade estimada de churn de aproximadamente **90,3%**.

Como essa probabilidade é superior ao ponto de corte de 50%, o modelo classificaria esse cliente como **propenso ao cancelamento da assinatura**.

Esse resultado demonstra como a regressão logística pode ser utilizada para identificar clientes com maior risco de evasão, permitindo que a empresa desenvolva ações preventivas e estratégias de retenção mais direcionadas.


Calculando a Probabilidade Estimada de duas formas diferentes
   - Com o log Odds
   - Com o Odds ratio
    

```python
from math import exp

meses_assinatura = 24
horas_assitidas = 32

# log odds
z_logOdds = (
    0.46498
    + 0.15890 * meses_assinatura
    - 0.06410 * horas_assitidas
)

# odds ratio
odds = (
    1.59198
    * (1.17222 ** meses_assinatura)
    * (0.93791 ** horas_assitidas)
)

# PROBABILIDADES
P_LO = exp(z_logOdds) / (1 + exp(z_logOdds))
P_OR = odds / (1 + odds)

print(P_LO)
print(P_OR)
```
